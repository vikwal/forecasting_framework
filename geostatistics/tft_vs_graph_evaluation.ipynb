# TFT vs. Graph-Modelle (DCRNN / MTGNN / WaveNet) — NWP & NWP+HIST

Vergleicht den nicht-graphenbasierten TFT-Baseline (`train_cl_tft_bc.py` / `get_test_results_tft_bc.py`) direkt mit den Graph-Modellen, für die GRID- (NWP-only) und GRID+HIST- (NWP + eigene Stationshistorie) Varianten. Gleicher Loader/gleiche Plot-Logik wie `fold_evaluation.ipynb`, erweitert um TFT.

TFT hat **keine BASE-Variante** (TFT selbst ist bereits die "kein Graph"-Referenz) — unter `Variant: BASE` erscheinen daher nur DCRNN/MTGNN/WaveNet.

```python
%matplotlib inline
import re
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import ipywidgets as widgets
from IPython.display import display

# Compat shim: this venv pins matplotlib==3.5.1, whose ColormapRegistry has no get_cmap()
# method. pandas>=2.1's Styler.background_gradient() (used in Plot 3) calls
# matplotlib.colormaps.get_cmap(...) unconditionally at render time, so it raises
# AttributeError the first time it actually renders non-empty data (previously masked
# because Plot 3's default state had no matching rows before the "run" dimension was
# added). Alias it to the subscript form ColormapRegistry already supports -- this only
# patches the imported module object in this notebook's kernel process, it does not touch
# the shared venv's installed packages (other jobs on this host depend on that install).
import matplotlib as _mpl
if not hasattr(_mpl.colormaps, "get_cmap"):
    _mpl.colormaps.get_cmap = lambda name: _mpl.colormaps[name]
```

```python
# ── Paths & mappings ────────────────────────────────────────────────────────
RESULTS_DIR = Path("../data/test_results")

MODEL_META = {
    "dcrnn_wind_dcrnn_base":         ("DCRNN",   "BASE"),
    "dcrnn_wind_dcrnn":              ("DCRNN",   "GRID"),
    "dcrnn_wind_dcrnn_nwp_hist":     ("DCRNN",   "GRID+HIST"),
    "mtgnn_wind_mtgnn":              ("MTGNN",   "BASE"),
    "mtgnn_wind_mtgnn_nwp":          ("MTGNN",   "GRID"),
    "mtgnn_wind_mtgnn_nwp_hist":     ("MTGNN",   "GRID+HIST"),
    "wavenet_wind_wavenet":          ("WaveNet", "BASE"),
    "wavenet_wind_wavenet_nwp":      ("WaveNet", "GRID"),
    "wavenet_wind_wavenet_nwp_hist": ("WaveNet", "GRID+HIST"),
    # TFT non-GNN baseline (get_test_results_tft_bc.py). No BASE variant — TFT itself
    # IS the "no graph" baseline; "base"/"hist" configs map to GRID/GRID+HIST (own
    # station history available yes/no), matching the DCRNN/MTGNN/WaveNet axis.
    "tft_wind_tft_base":             ("TFT",     "GRID"),
    "tft_wind_tft_hist":             ("TFT",     "GRID+HIST"),
    # NWP baselines (from evaluate_reference.py)
    "icon_d2":                   ("ICON-D2", "REF"),
    "ecmwf":                     ("ECMWF",   "REF"),
    # ── stdhp dry run (fixed standard hyperparameters, 3 SPATIAL folds -- see the "run"
    # dimension below). TFT itself has NO stdhp entry yet -- TFT training is still running
    # at the time of writing, see task notes; these are the graph-model stdhp results only.
    "stdhp_dcrnn_wind_dcrnn_base":     ("DCRNN",   "BASE"),
    "stdhp_dcrnn_wind_dcrnn":          ("DCRNN",   "GRID"),
    "stdhp_dcrnn_wind_dcrnn_nomeas":   ("DCRNN",   "GRID-NOMEAS"),
    "stdhp_dcrnn_wind_dcrnn_nograph":  ("DCRNN",   "GRID-NOGRAPH"),
    "stdhp_dcrnn_wind_dcrnn_nwp_hist": ("DCRNN",   "GRID+HIST"),
    "stdhp_mtgnn_wind_mtgnn":          ("MTGNN",   "BASE"),
    "stdhp_mtgnn_wind_mtgnn_nwp":      ("MTGNN",   "GRID"),
    "stdhp_mtgnn_wind_mtgnn_nwp_hist": ("MTGNN",   "GRID+HIST"),
    "stdhp_wavenet_wind_wavenet":      ("WaveNet", "BASE"),
    "stdhp_wavenet_wind_wavenet_nwp":  ("WaveNet", "GRID"),
}

# ── Run dimension (siehe fold_evaluation.ipynb fuer die ausfuehrliche Begruendung) ──────
# "hpo" = legacy/per-fold-HPO-best (temporal folds); "stdhp" = fixed standard hyperparameters
# (3 SPATIAL folds, identical time window); "ref" = evaluate_reference.py CSVs (wrong window
# for "stdhp", see _ref_available_for_run() below).
RUN_OPTIONS      = ["Hpo/Legacy", "Stdhp"]
RUN_LABEL_TO_KEY = {"Hpo/Legacy": "hpo", "Stdhp": "stdhp"}

def _run_of(prefix, model_name):
    if prefix.startswith("stdhp_"):
        return "stdhp"
    if model_name in ("ICON-D2", "ECMWF"):
        return "ref"
    return "hpo"

# Fold-Zeiträume sind SPLIT-abhängig für den "hpo"-Run. Der "stdhp"-Run ist räumlich gefoldet:
# alle 3 Folds decken dasselbe Zeitfenster ab (2024-08-01..2025-08-02), nur die 51 Zielstationen
# unterscheiden sich je Fold (siehe configs/spatial_folds.yaml).
FOLD_LABELS_VAL   = {0: "Aug–Nov 2024", 1: "Nov 2024–Mar 2025", 2: "Apr–Aug 2025"}
FOLD_LABELS_TEST  = {0: "Aug–Nov 2025", 1: "Dez 2025–März 2026"}
FOLD_LABELS_STDHP = {0: "Spatial fold 0", 1: "Spatial fold 1", 2: "Spatial fold 2"}
FOLD_LABELS_BY_SPLIT = {"val": FOLD_LABELS_VAL, "test": FOLD_LABELS_TEST}

# Backward-compat: FOLD_LABELS / FOLD_LABEL_TO_IDX referenzieren weiterhin Val/hpo.
FOLD_LABELS       = FOLD_LABELS_VAL
FOLD_LABEL_TO_IDX = {v: k for k, v in FOLD_LABELS_VAL.items()}

def _fold_labels_for(split, run="hpo"):
    if run == "stdhp":
        return FOLD_LABELS_STDHP
    return FOLD_LABELS_BY_SPLIT.get(str(split).lower(), FOLD_LABELS_VAL)

def _fold_idx_for(split, label, run="hpo"):
    """fold_label → fold-Index, split- UND run-abhängig (stdhp-Folds sind räumlich, nicht zeitlich)."""
    return {v: k for k, v in _fold_labels_for(split, run).items()}.get(label)

# Eval scripts produce: station_id, mae, rmse, r2, skill, skill_nwp, n_samples
# Notebook expects:     station,    val_mae, val_rmse, val_r2, skill_icond2, skill_ecmwf
_COL_RENAME = {
    "station_id": "station",
    "mae":        "val_mae",
    "rmse":       "val_rmse",
    "r2":         "val_r2",
    "skill_nwp":  "skill_icond2",
    "skill":      "skill_pers",
}

# ── Model CSV loader ────────────────────────────────────────────────────────
def load_all_csvs():
    frames = []
    for path in sorted(RESULTS_DIR.glob("*.csv")):
        m = re.match(r"(.+)_fold(\d+)$", path.stem)
        if not m:
            continue
        prefix, fold = m.group(1), int(m.group(2))
        # Test-Outputs heißen <prefix>_test_fold{n} → split erkennen + Suffix strippen.
        split = "val"
        if prefix.endswith("_test"):
            split, prefix = "test", prefix[:-len("_test")]
        if prefix not in MODEL_META:
            continue
        model_name, variant = MODEL_META[prefix]
        run = _run_of(prefix, model_name)
        df = pd.read_csv(path).rename(columns=_COL_RENAME)
        if "scenario"    not in df.columns: df["scenario"]    = "excl_val"
        if "seen"        not in df.columns: df["seen"]        = "n"
        if "skill_ecmwf" not in df.columns: df["skill_ecmwf"] = np.nan
        df["model"]      = model_name
        df["variant"]    = variant
        df["fold"]       = fold
        df["split"]      = split
        df["run"]        = run
        df["fold_label"] = _fold_labels_for(split, run).get(fold, "?")
        frames.append(df)
    if not frames:
        print("No CSV files found in", RESULTS_DIR)
        return pd.DataFrame()
    return pd.concat(frames, ignore_index=True)

# ── Reference metrics loader (legacy — superseded by evaluate_reference.py parquets) ────
def load_reference_metrics():
    path = RESULTS_DIR / "reference_metrics.csv"
    if not path.exists():
        return pd.DataFrame()
    df = pd.read_csv(path)
    df["fold_label"] = df["fold"].map(FOLD_LABELS)
    print(f"[reference_metrics.csv] {len(df):,} rows")
    return df

data    = load_all_csvs()
ref_all = load_reference_metrics()

print(f"Model data: {len(data):,} rows | "
      f"{data['model'].nunique() if not data.empty else 0} models | "
      f"splits: {sorted(data['split'].unique()) if not data.empty else []}")
if not data.empty:
    print("Models:", sorted(data["model"].unique()))
    print("Runs:  ", sorted(data["run"].unique()))
    for sp in sorted(data["split"].unique()):
        folds = sorted(data[data["split"] == sp]["fold"].unique())
        print(f"  split='{sp}': folds {folds}")
    if "TFT" not in data["model"].unique():
        print("Hinweis: keine TFT-Ergebnisse gefunden -- TFT-Training laeuft laut Auftrag noch.")

# ── Zeitfenster-Warnung fuer die evaluate_reference.py-CSVs ─────────────────────────────
# icon_d2_fold*.csv / ecmwf_fold*.csv decken 2025-08-01..2026-04-01 ab (evaluate_reference.py
# setzt die Untergrenze immer auf test_start). Sie werden hier mit split="val" geladen, weil ihr
# Dateiname nicht auf "_test" endet -- das ist ihr FENSTER aber nicht. Konkret heisst das:
#   * run="stdhp" (Val-Fenster 2024-08..2025-08): explizit deaktiviert, siehe _ref_available_for_run().
#   * run="hpo"   (auf der Platte nur split="test", Fold 1 = 2025-12..2026-04): die Referenzzeilen
#     tragen split="val" und joinen daher nie -- die ICON-D2/ECMWF-Balken bleiben stumm leer.
#     Selbst wenn sie jointen, waere das Fenster (Aug..Apr vs. Dez..Apr) verschieden.
# Es gibt also derzeit KEINE Konfiguration, in der diese CSVs eine gueltige Referenz sind.
# Korrekt gefensterte ICON-D2/Persistenz-Baselines: geostatistics/stdrun/make_stdhp_figures.py
# (aus den nwp_ref/pers_ref-Spalten der Modell-Parquets) bzw. evaluate_reference.py reparieren.
if not data.empty and (data["run"] == "ref").any():
    print("[WARN] icon_d2/ecmwf_fold*.csv geladen (run='ref', als split='val' getaggt), decken aber "
          "2025-08-01..2026-04-01 ab. Fuer run='stdhp' deaktiviert; fuer run='hpo' joinen sie nicht. "
          "Nicht als Referenz zitieren, bevor evaluate_reference.py neu gelaufen ist.")
```

```python
# ── Shared helpers ───────────────────────────────────────────────────────────

def ref_rmse_from_skill(df, ref_col):
    """Fallback: derive per-station reference RMSE from skill score."""
    denom = (1.0 - df[ref_col]).clip(lower=0.001, upper=0.999)
    return df["val_rmse"] / denom


def filter_model(df, variant, fold, scenario, seen, split="Val", run="Hpo/Legacy"):
    run_key = RUN_LABEL_TO_KEY[run]
    if "run" in df.columns:
        df = df[df["run"] == run_key]
    # Split zuerst (Val/Test haben getrennte Stationen UND — im hpo-Run — Fold-Zeiträume).
    if "split" in df.columns:
        df = df[df["split"] == str(split).lower()]
    df = df[df["variant"] == variant].copy()
    if fold != "Average":
        df = df[df["fold"] == _fold_idx_for(split, fold, run_key)]
    df = df[df["scenario"] == scenario]
    if seen == "Val stations only":
        df = df[df["seen"] == "n"]
    elif seen == "Train stations only":
        df = df[df["seen"] == "y"]
    return df


def filter_ref(df_ref, fold, seen, period="val", horizon=-1):
    """Filter reference_metrics to the aggregate val period for one fold."""
    if df_ref.empty:
        return df_ref
    df = df_ref[(df_ref["period"] == period) & (df_ref["horizon"] == horizon)].copy()
    if fold != "Average":
        df = df[df["fold"] == FOLD_LABEL_TO_IDX[fold]]
    if seen == "Val stations only":
        df = df[df["seen"] == "n"]
    elif seen == "Train stations only":
        df = df[df["seen"] == "y"]
    return df


def _bar_stats(series):
    return float(np.nanmean(series)), float(np.nanstd(series))


def _ref_available_for_run(run_key):
    """
    icon_d2_fold*.csv / ecmwf_fold*.csv (evaluate_reference.py) hard-code the evaluation
    window's lower bound to `test_start` regardless of --test-mode. For the 3 spatial folds
    this means they cover 2025-08-01..2026-04-01 (TEST period), not the "stdhp" run's VAL
    period 2024-08-01..2025-08-02 -- station sets match (51/51) but time windows do not.
    -> For run="stdhp", ICON-D2/ECMWF reference bars/lines fed from `data` are disabled.
       See geostatistics/stdrun/make_stdhp_figures.py for the correctly-windowed ICON-D2/
       Persistence baselines (derived from the stdhp raw parquets' own nwp_ref/pers_ref).
    """
    return run_key != "stdhp"


# ── Run + Split (+ Fold) widgets ─────────────────────────────────────────────
def _available_splits_for_run(run_key):
    if data.empty or "split" not in data.columns or "run" not in data.columns:
        return ["Val"]
    present = set(data[data["run"] == run_key]["split"].unique())
    opts = [s.capitalize() for s in ["val", "test"] if s in present]
    return opts or ["Val"]


def make_run_split_fold(style, layout, fold_desc="Fold:"):
    """Run + Split + Fold, dreifach gekoppelt (fuer Plots mit Fold-Selektor: 1, 2, 4, 5)."""
    w_run = widgets.Dropdown(options=RUN_OPTIONS, value="Stdhp",
                              description="Run:", style=style, layout=layout)
    run_key = RUN_LABEL_TO_KEY[w_run.value]
    splits  = _available_splits_for_run(run_key)
    w_split = widgets.Dropdown(options=splits, value=splits[0],
                               description="Split:", style=style, layout=layout)
    w_fold  = widgets.Dropdown(
        options=list(_fold_labels_for(splits[0], run_key).values()) + ["Average"],
        value="Average", description=fold_desc, style=style, layout=layout)

    def _on_split(change):
        rk = RUN_LABEL_TO_KEY[w_run.value]
        w_fold.options = list(_fold_labels_for(change["new"], rk).values()) + ["Average"]
        w_fold.value   = "Average"

    def _on_run(change):
        rk = RUN_LABEL_TO_KEY[change["new"]]
        new_splits = _available_splits_for_run(rk)
        if w_split.value in new_splits:
            w_split.options = new_splits
            w_fold.options  = list(_fold_labels_for(w_split.value, rk).values()) + ["Average"]
            w_fold.value    = "Average"
        else:
            w_split.options = new_splits
            w_split.value   = new_splits[0]   # triggert _on_split

    w_split.observe(_on_split, names="value")
    w_run.observe(_on_run, names="value")
    return w_run, w_split, w_fold


def make_run_split(style, layout):
    """Run + Split, ohne Fold-Selektor (fuer Plot 3 / 3b, die immer ueber alle Folds mitteln)."""
    w_run = widgets.Dropdown(options=RUN_OPTIONS, value="Stdhp",
                              description="Run:", style=style, layout=layout)
    run_key = RUN_LABEL_TO_KEY[w_run.value]
    splits  = _available_splits_for_run(run_key)
    w_split = widgets.Dropdown(options=splits, value=splits[0],
                               description="Split:", style=style, layout=layout)

    def _on_run(change):
        rk = RUN_LABEL_TO_KEY[change["new"]]
        new_splits = _available_splits_for_run(rk)
        w_split.options = new_splits
        if w_split.value not in new_splits:
            w_split.value = new_splits[0]

    w_run.observe(_on_run, names="value")
    return w_run, w_split
```

## Plot 1 — Model Comparison (Bar Chart)

```python
def plot_bar(metric, variant, fold, scenario, seen, split, run):
    run_key = RUN_LABEL_TO_KEY[run]
    df = filter_model(data, variant, fold, scenario, seen, split, run)

    if df.empty:
        print("No data available for the current selection.")
        return

    MODELS = ["DCRNN", "MTGNN", "WaveNet", "TFT"]
    bars, vals, errs, colors = [], [], [], []

    def _add_model(col):
        for m in MODELS:
            s = df[df["model"] == m][col].dropna()
            if s.empty: continue
            bars.append(m); v, e = _bar_stats(s); vals.append(v); errs.append(e)
            colors.append("steelblue")

    def _add_ref_from_data(col):
        """ICON-D2 / ECMWF bars from evaluate_reference.py CSVs (loaded into data).
        Disabled for run='stdhp' -- see _ref_available_for_run()."""
        if not _ref_available_for_run(run_key):
            return
        ref_df = data[data["model"].isin(["ICON-D2", "ECMWF"]) & (data["split"] == split.lower())]
        if fold != "Average":
            ref_df = ref_df[ref_df["fold"] == _fold_idx_for(split, fold, run_key)]
        for ref_model, color in [("ICON-D2", "#888888"), ("ECMWF", "#555555")]:
            s = ref_df[ref_df["model"] == ref_model][col].dropna()
            if s.empty: continue
            bars.append(ref_model); v, e = _bar_stats(s); vals.append(v); errs.append(e)
            colors.append(color)

    if metric == "RMSE":
        _add_model("val_rmse");          _add_ref_from_data("val_rmse"); ylabel = "RMSE (m/s)"
    elif metric == "MAE":
        _add_model("val_mae");           _add_ref_from_data("val_mae");  ylabel = "MAE (m/s)"
    elif metric == "R²":
        _add_model("val_r2");            _add_ref_from_data("val_r2");   ylabel = "R²"
    elif metric in ("Skill ICON-D2", "Skill ECMWF"):
        skill_col = "skill_icond2" if "ICON-D2" in metric else "skill_ecmwf"
        _add_model(skill_col)
        if _ref_available_for_run(run_key):
            bars  += ["ICON-D2", "ECMWF"]; vals  += [0.0, 0.0]
            errs  += [0.0, 0.0];           colors += ["#888888", "#555555"]
        ylabel = f"Skill vs. {'ICON-D2' if 'ICON-D2' in metric else 'ECMWF'}"

    if not bars:
        print("No data available for the current selection.")
        return

    fig, ax = plt.subplots(figsize=(max(6, len(bars) * 1.3), 5))
    x = np.arange(len(bars))
    ax.bar(x, vals, yerr=errs, capsize=5, color=colors,
           edgecolor="white", linewidth=0.5,
           error_kw={"elinewidth": 1.5, "alpha": 0.7})

    if metric == "R²":
        ymin = min((v for v in vals if np.isfinite(v)), default=-0.1) - 0.05
        ax.set_ylim(min(ymin, -0.05), 1.12)
        ax.axhline(0.0, color="red", linestyle="--", linewidth=1.5, alpha=0.8)
        ax.axhline(1.0, color="red", linestyle="--", linewidth=1.5, alpha=0.8)
    elif metric in ("Skill ICON-D2", "Skill ECMWF"):
        ax.axhline(0.0, color="gray", linestyle="--", linewidth=1.0, alpha=0.8)

    ax.set_xticks(x)
    ax.set_xticklabels(bars, fontsize=12)
    ax.tick_params(axis="y", labelsize=12)
    ax.set_ylabel(ylabel, fontsize=12)
    ax.axhline(0, color="gray", linewidth=0.6)

    if not _ref_available_for_run(run_key):
        ax.text(0.02, 0.98,
                "ICON-D2/ECMWF disabled for Run=Stdhp\n(wrong time window in evaluate_reference.py\n"
                "CSVs here -- see stdrun/make_stdhp_figures.py)",
                transform=ax.transAxes, ha="left", va="top", fontsize=8.5, color="firebrick",
                bbox=dict(boxstyle="round,pad=0.3", fc="#fff3f3", ec="firebrick", alpha=0.9))

    fold_text = fold if fold != "Average" else "Average (all folds)"
    ax.text(
        0.98, 0.98,
        f"Run: {run}\nSplit: {split}\nFold: {fold_text}\nScenario: {scenario}\nStations: {seen}",
        transform=ax.transAxes, ha="right", va="top", fontsize=10,
        bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="#ccc", alpha=0.85),
    )
    plt.tight_layout()
    plt.show(); plt.close()


_style  = {"description_width": "70px"}
_layout = widgets.Layout(width="200px")

w1_metric   = widgets.Dropdown(
    options=["RMSE", "MAE", "R²", "Skill ICON-D2", "Skill ECMWF"],
    value="RMSE", description="Metric:", style=_style, layout=_layout)
w1_variant  = widgets.Dropdown(
    options=["BASE", "GRID", "GRID-NOMEAS", "GRID-NOGRAPH", "GRID+HIST"],
    value="BASE", description="Variant:", style=_style, layout=_layout)
w1_run, w1_split, w1_fold = make_run_split_fold(_style, _layout)
w1_scenario = widgets.Dropdown(
    options=["excl_val", "incl_val"],
    value="excl_val", description="Scenario:", style=_style, layout=_layout)
w1_seen     = widgets.Dropdown(
    options=["Val stations only", "Train stations only", "Both combined"],
    value="Val stations only", description="Stations:", style=_style, layout=_layout)

out1 = widgets.interactive_output(
    plot_bar,
    {"metric": w1_metric, "variant": w1_variant, "fold": w1_fold,
     "scenario": w1_scenario, "seen": w1_seen, "split": w1_split, "run": w1_run},
)
display(widgets.HBox([w1_run, w1_split, w1_metric, w1_variant, w1_fold, w1_scenario, w1_seen]), out1)
```

## Plot 2 — Station-Level Scatter (Model vs. Reference)

```python
# Metric column mapping: (model_col, xlabel_unit, higher_is_better)
_METRIC_MAP = {
    "RMSE": ("val_rmse", "RMSE (m/s)", False),
    "MAE":  ("val_mae",  "MAE (m/s)",  False),
    "R²":   ("val_r2",   "R²",         True),
}


def _get_ref_per_station(model_col, fold, split, run_key):
    """Per-station reference metrics from evaluate_reference.py CSVs (in data), split-aware.
    Disabled for run='stdhp' -- see _ref_available_for_run()."""
    result = {"ICON-D2": pd.DataFrame(), "ECMWF": pd.DataFrame()}
    if not _ref_available_for_run(run_key):
        return result
    for ref_mdl in ["ICON-D2", "ECMWF"]:
        r = data[(data["model"] == ref_mdl) & (data["split"] == split.lower())][
            ["station", "fold", model_col]].copy()
        if r.empty:
            continue
        if fold == "Average":
            agg = r.groupby("station")[model_col].mean().reset_index()
        else:
            agg = r[r["fold"] == _fold_idx_for(split, fold, run_key)][["station", model_col]].copy()
        result[ref_mdl] = agg.rename(columns={model_col: "ref"})
    return result


def _scatter_panel(ax, x, y, metric_name, xlabel):
    higher_better = _METRIC_MAP[metric_name][2]
    better = (y > x) if higher_better else (y < x)
    worse  = ~better

    ax.scatter(x[better], y[better], c="#2ca02c", alpha=0.55, s=28, label=f"Better  (n={better.sum()})")
    ax.scatter(x[worse],  y[worse],  c="#d62728", alpha=0.55, s=28, label=f"Worse   (n={worse.sum()})")

    finite = np.concatenate([x, y])
    finite = finite[np.isfinite(finite)]
    if len(finite) == 0:
        return

    if higher_better:
        lo = min(np.nanmin(finite) - 0.05, 0.0)
        hi = max(np.nanmax(finite) + 0.05, 1.0)
        ax.plot([lo, hi], [lo, hi], "--", color="#aaa", linewidth=1.2, zorder=0)
        ax.set_xlim(lo, hi); ax.set_ylim(lo, hi)
    else:
        lim_max = max(np.nanpercentile(finite, 99) * 1.08, 0.3)
        ax.plot([0, lim_max], [0, lim_max], "--", color="#aaa", linewidth=1.2, zorder=0)
        ax.set_xlim(0, lim_max); ax.set_ylim(0, lim_max)

    ax.set_xlabel(xlabel, fontsize=12)
    ax.set_ylabel(f"Model {metric_name}", fontsize=12)
    ax.tick_params(labelsize=12)
    ax.legend(fontsize=10, loc="upper left", framealpha=0.85)


def plot_scatter(metric, model, variant, fold, scenario, seen, split, run):
    run_key = RUN_LABEL_TO_KEY[run]
    model_col, metric_unit, _ = _METRIC_MAP[metric]

    df   = filter_model(data, variant, fold, scenario, seen, split, run)
    df_m = df[df["model"] == model].copy() if not df.empty else pd.DataFrame()
    if df_m.empty:
        print(f"No data for model '{model}' with the current selection.")
        return

    if fold == "Average":
        m_agg = df_m.groupby("station")[model_col].mean().reset_index()
    else:
        m_agg = df_m[df_m["fold"] == _fold_idx_for(split, fold, run_key)][["station", model_col]].copy()

    refs = _get_ref_per_station(model_col, fold, split, run_key)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    for ax, ref_name in [(ax1, "ICON-D2"), (ax2, "ECMWF")]:
        ref_df = refs.get(ref_name, pd.DataFrame())
        ax.set_title(f"Model vs {ref_name}", fontsize=12)

        if not _ref_available_for_run(run_key):
            ax.text(0.5, 0.5, f"{ref_name} disabled for Run=Stdhp\n(wrong time window in\n"
                               "evaluate_reference.py CSVs --\nsee stdrun/make_stdhp_figures.py)",
                    ha="center", va="center", transform=ax.transAxes, fontsize=10, color="firebrick")
            continue

        if ref_df.empty:
            ax.text(0.5, 0.5, f"No {ref_name} reference data\n(run evaluate_reference.py)",
                    ha="center", va="center", transform=ax.transAxes, fontsize=11)
            continue

        merged = m_agg.merge(ref_df, on="station", how="inner").dropna(subset=[model_col, "ref"])
        if len(merged) < 2:
            ax.text(0.5, 0.5, f"Insufficient data for {ref_name}",
                    ha="center", va="center", transform=ax.transAxes, fontsize=11)
            continue

        x  = merged["ref"].values
        y  = merged[model_col].values
        ok = np.isfinite(x) & np.isfinite(y)
        if ok.sum() < 2:
            ax.text(0.5, 0.5, "No finite pairs", ha="center", va="center",
                    transform=ax.transAxes, fontsize=11)
            continue

        _scatter_panel(ax, x[ok], y[ok], metric, f"{ref_name} {metric_unit}")

    fold_text = fold if fold != "Average" else "Average (all folds)"
    fig.text(0.5, -0.02,
             f"Run: {run} | Split: {split} | Model: {model} | Variant: {variant} | Metric: {metric} | "
             f"Fold: {fold_text} | Scenario: {scenario} | Stations: {seen}",
             ha="center", fontsize=10, color="#555")
    plt.tight_layout()
    plt.show(); plt.close()


_available_models = [m for m in ["DCRNN", "MTGNN", "WaveNet", "TFT"]
                     if not data.empty and m in data["model"].values] or ["DCRNN", "MTGNN", "WaveNet", "TFT"]

w2_metric   = widgets.Dropdown(options=["RMSE", "MAE", "R²"],
                               value="RMSE", description="Metric:", style=_style, layout=_layout)
w2_model    = widgets.Dropdown(options=_available_models, value=_available_models[0],
                               description="Model:", style=_style, layout=_layout)
w2_variant  = widgets.Dropdown(options=["BASE", "GRID", "GRID-NOMEAS", "GRID-NOGRAPH", "GRID+HIST"],
                               value="BASE", description="Variant:", style=_style, layout=_layout)
w2_run, w2_split, w2_fold = make_run_split_fold(_style, _layout)
w2_scenario = widgets.Dropdown(options=["excl_val", "incl_val"],
                               value="excl_val", description="Scenario:", style=_style, layout=_layout)
w2_seen     = widgets.Dropdown(options=["Val stations only", "Train stations only", "Both combined"],
                               value="Val stations only", description="Stations:", style=_style, layout=_layout)

out2 = widgets.interactive_output(
    plot_scatter,
    {"metric": w2_metric, "model": w2_model, "variant": w2_variant,
     "fold": w2_fold, "scenario": w2_scenario, "seen": w2_seen, "split": w2_split, "run": w2_run},
)
display(widgets.HBox([w2_run, w2_split, w2_metric, w2_model, w2_variant, w2_fold, w2_scenario, w2_seen]), out2)
```

```python
# ── Plot 3: Summary table ────────────────────────────────────────────────────
_SEEN_TO_FILTER = {
    "Train": "Train stations only",
    "Val":   "Val stations only",
    "Both":  "Both combined",
}
_SEEN_TO_DATA = {"Train": "y", "Val": "n", "Both": None}


def show_summary_table(variant, seen, split, run):
    run_key = RUN_LABEL_TO_KEY[run]
    seen_val = _SEEN_TO_DATA[seen]
    sp = split.lower()

    mask = (data["run"] == run_key) & (data["split"] == sp) & data["model"].isin(["DCRNN", "MTGNN"]) & (data["variant"] == variant)
    if seen_val is not None:
        mask &= data["seen"] == seen_val
    df_filt = data[mask]

    agg = (
        df_filt.groupby("model")[["val_r2", "val_rmse", "skill_icond2", "skill_ecmwf"]]
        .mean()
        .reindex(["DCRNN", "MTGNN"])
        .rename(columns={
            "val_r2":        "R²",
            "val_rmse":      "RMSE",
            "skill_icond2":  "Skill ICON-D2",
            "skill_ecmwf":   "Skill ECMWF",
        })
    )

    ref_note = ""
    if _ref_available_for_run(run_key):
        for ref_mdl, label in [("ICON-D2", "ICON-D2"), ("ECMWF", "ECMWF")]:
            ref_sub = data[(data["model"] == ref_mdl) & (data["split"] == sp)]
            if not ref_sub.empty:
                agg[f"R² {label}"]   = ref_sub["val_r2"].mean()
                agg[f"RMSE {label}"] = ref_sub["val_rmse"].mean()
    else:
        ref_note = " | ICON-D2/ECMWF columns disabled for Run=Stdhp (wrong time window, see make_stdhp_figures.py)"

    grad_up   = [c for c in ["R²", "Skill ICON-D2", "Skill ECMWF",
                              "R² ICON-D2", "R² ECMWF"] if c in agg.columns]
    grad_down = [c for c in ["RMSE", "RMSE ICON-D2", "RMSE ECMWF"] if c in agg.columns]

    styled = (
        agg.style
        .format("{:.4f}", na_rep="—")
        .background_gradient(subset=grad_up,   cmap="RdYlGn",   axis=0)
        .background_gradient(subset=grad_down, cmap="RdYlGn_r", axis=0)
        .set_caption(f"Run: {run} | Split: {split} | Variant: {variant} | Seen: {seen} | "
                     f"Ø über alle Folds & Stationen{ref_note}")
    )
    display(styled)


w3_run, w3_split = make_run_split(_style, _layout)
w3_variant = widgets.Dropdown(
    options=["BASE", "GRID", "GRID-NOMEAS", "GRID-NOGRAPH", "GRID+HIST"],
    value="BASE", description="Variant:", style=_style, layout=_layout)
w3_seen = widgets.Dropdown(
    options=["Train", "Val", "Both"],
    value="Val", description="Seen:", style=_style, layout=_layout)

out3 = widgets.interactive_output(
    show_summary_table, {"variant": w3_variant, "seen": w3_seen, "split": w3_split, "run": w3_run})
display(widgets.HBox([w3_run, w3_split, w3_variant, w3_seen]), out3)
```

```python
# ── Plot 3b: Comparison bar chart — ICON-D2 vs NWP only vs NWP + Hist ───────
def _improvement_bracket(ax, x0, x1, v0, v1, higher_better, lift=0.0,
                          color="dimgray", fontsize=11):
    """
    Significance bracket placed in the space between bar tops and ylim.
    `lift` = 0.0 places the bracket low in that space, 1.0 places it high.
    """
    if np.isnan(v0) or np.isnan(v1):
        return
    pct  = (v1 - v0) / abs(v0) * 100 if higher_better else (v0 - v1) / abs(v0) * 100
    sign = "+" if pct >= 0 else ""

    _, ymax    = ax.get_ylim()
    headroom   = ymax - max(v0, v1)        # space above the taller bar
    tick       = headroom * 0.10
    y_top      = max(v0, v1) + headroom * (0.28 + lift * 0.40)

    ax.plot([x0, x0, x1, x1],
            [v0 + tick, y_top, y_top, v1 + tick],
            color=color, lw=1.2)
    ax.text((x0 + x1) / 2, y_top + headroom * 0.05,
            f"{sign}{pct:.1f}%", ha="center", va="bottom",
            fontsize=fontsize, color=color, fontweight="bold")


def show_comparison_plot(seen, split, run):
    run_key = RUN_LABEL_TO_KEY[run]
    seen_val = _SEEN_TO_DATA[seen]
    sp = split.lower()
    ref_ok = _ref_available_for_run(run_key)

    # --- Reference (ICON-D2) from evaluate_reference.py CSVs (loaded into `data`), split-aware ---
    def _ref_mean(col):
        if not ref_ok:
            return np.nan
        m = (data["model"] == "ICON-D2") & (data["split"] == sp)
        if seen_val is not None:
            m &= data["seen"] == seen_val
        s = data[m][col]
        return float(s.mean()) if not s.empty else np.nan

    r2_ref   = _ref_mean("val_r2")
    rmse_ref = _ref_mean("val_rmse")

    # --- Model means (DCRNN GRID / MTGNN GRID+HIST, fixed) ---
    def _mean(model, variant, col):
        m = (data["run"] == run_key) & (data["model"] == model) & (data["variant"] == variant) & (data["split"] == sp)
        if seen_val is not None:
            m &= data["seen"] == seen_val
        s = data[m][col]
        return float(s.mean()) if not s.empty else np.nan

    r2_vals   = [r2_ref,
                 _mean("DCRNN", "GRID",      "val_r2"),
                 _mean("MTGNN", "GRID+HIST", "val_r2")]
    rmse_vals = [rmse_ref,
                 _mean("DCRNN", "GRID",      "val_rmse"),
                 _mean("MTGNN", "GRID+HIST", "val_rmse")]

    labels = ["ICON-D2", "NWP only", "NWP + Hist"]
    green  = "#2ca02c"
    x      = np.arange(len(labels))
    bar_w  = 0.6

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    # ── Left: R² ─────────────────────────────────────────────────────────────
    ax1.bar(x, r2_vals, color=green, alpha=0.85, width=bar_w, zorder=3)
    ax1.set_xticks(x)
    ax1.set_xticklabels(labels, fontsize=12)
    ax1.set_ylabel("R²", fontsize=12)
    ax1.tick_params(labelsize=12)
    ax1.set_ylim(0, 1.0)
    ax1.set_yticks(np.arange(0, 1.1, 0.2))
    ax1.set_yticklabels([f"{v:.1f}" for v in np.arange(0, 1.1, 0.2)], fontsize=12)

    # Dashed container outlines (0 → 1) per bar
    for xi in x:
        rect = mpatches.Rectangle(
            (xi - bar_w / 2, 0), bar_w, 1.0,
            fill=False, linestyle="--", edgecolor="gray",
            linewidth=0.9, alpha=0.65, zorder=4,
        )
        ax1.add_patch(rect)

    # Brackets inside the headroom between bar tops and y=1
    _improvement_bracket(ax1, 0, 1, r2_vals[0], r2_vals[1], higher_better=True, lift=0.0)
    _improvement_bracket(ax1, 0, 2, r2_vals[0], r2_vals[2], higher_better=True, lift=1.0)

    ax1.set_title("R²", fontsize=12)

    # ── Right: RMSE ──────────────────────────────────────────────────────────
    ax2.bar(x, rmse_vals, color=green, alpha=0.85, width=bar_w, zorder=3)
    ax2.set_xticks(x)
    ax2.set_xticklabels(labels, fontsize=12)
    ax2.set_ylabel("RMSE (m/s)", fontsize=12)
    ax2.tick_params(labelsize=12)

    max_rmse = max((v for v in rmse_vals if not np.isnan(v)), default=1.0)
    ax2.set_ylim(0, max_rmse * 1.45)

    _improvement_bracket(ax2, 0, 1, rmse_vals[0], rmse_vals[1], higher_better=False, lift=0.0)
    _improvement_bracket(ax2, 0, 2, rmse_vals[0], rmse_vals[2], higher_better=False, lift=1.0)

    ax2.set_title("RMSE", fontsize=12)

    if not ref_ok:
        fig.text(0.5, -0.06,
                 "ICON-D2 bar disabled for Run=Stdhp (evaluate_reference.py CSVs cover the wrong "
                 "time window here) -- see geostatistics/stdrun/make_stdhp_figures.py",
                 ha="center", fontsize=9, color="firebrick")

    fig.suptitle(f"NWP baseline comparison — Run: {run} | Split: {split} | Stations: {seen}", fontsize=12, y=1.01)
    plt.tight_layout()
    plt.show()
    plt.close()


out_cmp = widgets.interactive_output(show_comparison_plot, {"seen": w3_seen, "split": w3_split, "run": w3_run})
display(out_cmp)
```

```python
# ── Raw predictions loader ────────────────────────────────────────────────────
RAW_DIR = Path("../data/raw_preds")

def load_raw_preds():
    frames = []
    for path in sorted(RAW_DIR.glob("*_raw.parquet")):
        # Filename: {prefix}_fold{n}_raw.parquet  (Test: {prefix}_test_fold{n}_raw.parquet)
        stem = path.stem[:-4]  # strip _raw suffix (4 chars)
        m = re.match(r"(.+)_fold(\d+)$", stem)
        if not m:
            continue
        prefix = m.group(1)
        fold   = int(m.group(2))
        split  = "val"
        if prefix.endswith("_test"):
            split, prefix = "test", prefix[:-len("_test")]
        if prefix not in MODEL_META:
            continue
        model_name, variant = MODEL_META[prefix]
        run = _run_of(prefix, model_name)
        df = pd.read_parquet(path)
        df["station_id"] = df["station_id"].astype(str).str.zfill(5)
        df["model"]      = model_name
        df["variant"]    = variant
        df["fold"]       = fold
        df["split"]      = split
        df["run"]        = run
        df["fold_label"] = _fold_labels_for(split, run).get(fold, "?")
        df["hour"]       = pd.to_datetime(df["valid_time"]).dt.hour
        df["month"]      = pd.to_datetime(df["valid_time"]).dt.month
        # run_hour = Uhrzeit, zu der die Prognose erstellt wurde (ICON-D2-Lauf: 6/9/12/15)
        df["run_hour"]   = pd.to_datetime(df["run_time"]).dt.hour
        frames.append(df)
    if not frames:
        print(f"Keine Rohdaten in {RAW_DIR} — bitte get_test_results_*.py ausführen.")
        return pd.DataFrame()
    return pd.concat(frames, ignore_index=True)

raw = load_raw_preds()
print(f"Raw predictions: {len(raw):,} Zeilen | {raw['model'].nunique() if not raw.empty else 0} Modelle")
if not raw.empty:
    print("Splits:", sorted(raw["split"].unique()),
          "| Runs:", sorted(raw["run"].unique()),
          "| Prognose-Laufzeiten (run_hour):", sorted(raw["run_hour"].unique()))
```

## Plot 4 — Fehler nach Prognosehorizont

```python
# ── Shared helper for raw stratification ─────────────────────────────────────
_MODEL_COLORS = {
    "DCRNN":       "steelblue",
    "MTGNN":       "darkorange",
    "WaveNet":     "forestgreen",
    "TFT":         "crimson",
    "ICON-D2":     "#888888",
    "Persistence": "#c44e52",
}

def _metric_from_err(pred, gt, metric):
    err = pred - gt
    if metric == "RMSE":
        return float(np.sqrt((err ** 2).mean()))
    elif metric == "MAE":
        return float(err.abs().mean())
    elif metric == "R²":
        ss_res = float((err ** 2).sum())
        ss_tot = float(((gt - gt.mean()) ** 2).sum())
        return 1.0 - ss_res / ss_tot if ss_tot > 0 else np.nan
    raise ValueError(f"Unknown metric {metric}")

def _compute_metric(grp, metric):
    return _metric_from_err(grp["pred"], grp["gt"], metric)

def _compute_metric_ref(grp, metric, col):
    return _metric_from_err(grp[col], grp["gt"], metric)

def _metric_ylabel(metric):
    return metric if metric == "R²" else f"{metric} (m/s)"


def plot_horizon(metric, variant, fold, run_hour, split, run):
    if raw.empty:
        print("Keine Rohdaten vorhanden.")
        return
    run_key = RUN_LABEL_TO_KEY[run]

    # nur run_key in {"hpo","stdhp"} -- schliesst die separat getrackten icon_d2/ecmwf-Zeilen
    # (run="ref", falsches Zeitfenster fuer die raeumlichen Folds) explizit aus. ICON-D2/
    # Persistenz kommen hier ausschliesslich aus den modell-eigenen nwp_ref/pers_ref-Spalten.
    df = raw[(raw["split"] == split.lower()) & (raw["run"] == run_key)].copy()
    if variant != "All":
        df = df[df["variant"] == variant]
    if fold != "Average":
        df = df[df["fold"] == _fold_idx_for(split, fold, run_key)]
    if run_hour != "Alle":
        df = df[df["run_hour"] == int(run_hour)]
    if df.empty:
        print("Keine Daten für diese Auswahl.")
        return

    fig, ax = plt.subplots(figsize=(12, 5))

    for mdl in ["DCRNN", "MTGNN", "WaveNet", "TFT"]:
        dm = df[df["model"] == mdl]
        if dm.empty:
            continue
        vals = dm.groupby("horizon").apply(lambda g: _compute_metric(g, metric))
        ax.plot(vals.index.to_numpy(), vals.to_numpy(), label=mdl,
                color=_MODEL_COLORS.get(mdl), linewidth=2, marker=".", markersize=4)

    for ref_col, ref_label, color_key in [("nwp_ref", "ICON-D2 (nwp_ref)", "ICON-D2"),
                                           ("pers_ref", "Persistence (pers_ref)", "Persistence")]:
        if ref_col not in df.columns:
            continue
        ref_vals = df.groupby("horizon").apply(lambda g: _compute_metric_ref(g, metric, ref_col))
        ax.plot(ref_vals.index.to_numpy(), ref_vals.to_numpy(), label=ref_label,
                color=_MODEL_COLORS[color_key], linewidth=1.5, linestyle="--")

    ax.set_xlabel("Prognosehorizont (Stunden)", fontsize=12)
    ax.set_ylabel(_metric_ylabel(metric), fontsize=12)
    ax.legend(fontsize=11, loc="upper left")
    ax.grid(axis="y", alpha=0.4)
    fold_text = fold if fold != "Average" else "Ø alle Folds"
    run_text  = "alle Läufe" if run_hour == "Alle" else f"{run_hour}-Uhr-Lauf"
    ax.set_title(f"{metric} nach Prognosehorizont | Run: {run} | Split: {split} | Variante: {variant} | "
                 f"{fold_text} | {run_text}", fontsize=13)
    plt.tight_layout()
    plt.show(); plt.close()


_style4  = {"description_width": "90px"}
_layout4 = widgets.Layout(width="210px")

_run_hour_opts = ["Alle"] + (sorted(int(h) for h in raw["run_hour"].unique()) if not raw.empty else [6, 9, 12, 15])

w4_metric   = widgets.Dropdown(options=["RMSE", "MAE", "R²"], value="RMSE",
                               description="Metrik:", style=_style4, layout=_layout4)
w4_variant  = widgets.Dropdown(options=["All", "BASE", "GRID", "GRID-NOMEAS", "GRID-NOGRAPH", "GRID+HIST"],
                               value="BASE", description="Variante:", style=_style4, layout=_layout4)
w4_run, w4_split, w4_fold = make_run_split_fold(_style4, _layout4)
w4_runhour  = widgets.Dropdown(options=_run_hour_opts, value="Alle",
                               description="Prognose-Lauf:", style=_style4, layout=_layout4)

out4 = widgets.interactive_output(
    plot_horizon, {"metric": w4_metric, "variant": w4_variant, "fold": w4_fold,
                   "run_hour": w4_runhour, "split": w4_split, "run": w4_run}
)
display(widgets.HBox([w4_run, w4_split, w4_metric, w4_variant, w4_fold, w4_runhour]), out4)
```

## Plot 5 — Fehler nach Windgeschwindigkeit / Tagesstunde / Monat / Horizont (Boxplots)

```python
# ── Plot 5 helper ────────────────────────────────────────────────────────────
_WS_BINS = list(range(0, 22, 2))   # [0,2,4,...,20] = 11 edges
# Intervall-Labels: linke Grenze inklusiv [, rechte exklusiv ) — right=False in pd.cut.
# 10 reguläre Intervalle + offenes Endintervall = 11 Labels (passt zu bins+[inf]).
_WS_LABELS = [f"[{_WS_BINS[i]}, {_WS_BINS[i+1]})" for i in range(len(_WS_BINS) - 1)] + ["[20, ∞)"]

_X_OPTS = {
    "Prognosehorizont": ("horizon",  "Prognosehorizont (h)"),
    "Tagesstunde":      ("hour",     "Stunde (UTC)"),
    "Monat":            ("month",    "Monat"),
    "WS-Klasse (m/s)":  ("ws_bin",  "Windgeschwindigkeit (m/s)  [links inkl., rechts exkl.)"),
}


def plot_stratified(x_axis, metric, variant, fold, plot_type, split, run):
    if raw.empty:
        print("Keine Rohdaten vorhanden.")
        return
    run_key = RUN_LABEL_TO_KEY[run]

    df = raw[(raw["split"] == split.lower()) & (raw["run"] == run_key)].copy()
    if variant != "All":
        df = df[df["variant"] == variant]
    if fold != "Average":
        df = df[df["fold"] == _fold_idx_for(split, fold, run_key)]
    if df.empty:
        print("Keine Daten für diese Auswahl.")
        return

    x_col, xlabel = _X_OPTS[x_axis]

    if x_col == "ws_bin":
        df = df.copy()
        df["ws_bin"] = pd.cut(df["gt"], bins=_WS_BINS + [np.inf], labels=_WS_LABELS, right=False)
        df = df.dropna(subset=["ws_bin"])

    models = [m for m in ["DCRNN", "MTGNN", "WaveNet", "TFT"] if m in df["model"].unique()]
    if not models:
        print("Keine Modell-Daten für diese Auswahl.")
        return

    if plot_type == "Linie":
        fig, ax = plt.subplots(figsize=(12, 5))
        for mdl in models:
            dm = df[df["model"] == mdl]
            vals = dm.groupby(x_col, observed=True).apply(lambda g: _compute_metric(g, metric))
            xs = np.arange(len(vals)) if x_col == "ws_bin" else vals.index.to_numpy()
            ax.plot(xs, vals.to_numpy(), label=mdl, color=_MODEL_COLORS.get(mdl), linewidth=2, marker=".")
        if x_col == "ws_bin":
            ax.set_xticks(range(len(_WS_LABELS)))
            ax.set_xticklabels(_WS_LABELS, rotation=45, ha="right")
        ax.set_xlabel(xlabel, fontsize=12)
        ax.set_ylabel(f"{metric} (m/s)", fontsize=12)
        ax.legend(fontsize=11)
        ax.grid(axis="y", alpha=0.4)

    else:  # Boxplot — Ausreißer (Flier) ausgeblendet, sonst überladen der Plot
        n_models = len(models)
        fig, axes = plt.subplots(1, n_models, figsize=(6 * n_models, 5), sharey=True)
        if n_models == 1:
            axes = [axes]
        for ax, mdl in zip(axes, models):
            dm = df[df["model"] == mdl].copy()
            dm["abs_err"] = (dm["pred"] - dm["gt"]).abs()
            order = _WS_LABELS if x_col == "ws_bin" else sorted(dm[x_col].unique())
            groups = [dm[dm[x_col] == v]["abs_err"].dropna().values for v in order]
            ax.boxplot(groups, labels=[str(v) for v in order],
                       showfliers=False,
                       patch_artist=True,
                       boxprops=dict(facecolor=_MODEL_COLORS.get(mdl, "steelblue"), alpha=0.6))
            ax.set_title(mdl, fontsize=12)
            ax.set_xlabel(xlabel, fontsize=11)
            ax.tick_params(axis="x", rotation=45)
            if ax == axes[0]:
                ax.set_ylabel("|Fehler| (m/s)", fontsize=11)
            ax.grid(axis="y", alpha=0.3)

    fold_text = fold if fold != "Average" else "Ø alle Folds"
    fig.suptitle(
        f"{x_axis} | Run: {run} | Split: {split} | Metrik: {metric} | Variante: {variant} | {fold_text}",
        fontsize=13, y=1.02
    )
    plt.tight_layout()
    plt.show(); plt.close()


_style5  = {"description_width": "90px"}
_layout5 = widgets.Layout(width="220px")

w5_xaxis   = widgets.Dropdown(options=list(_X_OPTS.keys()), value="WS-Klasse (m/s)",
                               description="X-Achse:", style=_style5, layout=_layout5)
w5_metric  = widgets.Dropdown(options=["RMSE", "MAE"], value="RMSE",
                               description="Metrik:", style=_style5, layout=_layout5)
w5_variant = widgets.Dropdown(options=["All", "BASE", "GRID", "GRID-NOMEAS", "GRID-NOGRAPH", "GRID+HIST"],
                               value="BASE", description="Variante:", style=_style5, layout=_layout5)
w5_run, w5_split, w5_fold = make_run_split_fold(_style5, _layout5)
w5_type    = widgets.Dropdown(options=["Boxplot", "Linie"], value="Boxplot",
                               description="Plottyp:", style=_style5, layout=_layout5)

out5 = widgets.interactive_output(
    plot_stratified,
    {"x_axis": w5_xaxis, "metric": w5_metric, "variant": w5_variant,
     "fold": w5_fold, "plot_type": w5_type, "split": w5_split, "run": w5_run}
)
display(widgets.HBox([w5_xaxis, w5_run, w5_split, w5_metric, w5_variant, w5_fold, w5_type]), out5)
```
